```python
import numpy as np
arr = np.array([2,3,4,6,7,8,9,12,13,17,23,25,27,34,37,201])
arr
```

```python
q1 = np.percentile(arr, 25)
q3 = np.percentile(arr, 75)
```

```python
IQR = q3 - q1
IQR
```

```python
upper_fench = q3 + 1.5*(IQR)
lower_fench = q1 - 1.5*(IQR)
lower_fench
```

```python
l = []
for i in arr:
    if i <= upper_fench and i>=lower_fench:
        l.append(i)
print(l)
arr2 = np.array(l)
arr2
```

```python
import seaborn as sns

sns.boxplot(x=arr)
```
