```python
X= df.drop(columns=['price'], axis = 1)
y = df['price']
```

```python
df.columns
```

```python
X_one_encoded = pd.get_dummies(X, columns=['model', 'transmission', 'fuelType'], drop_first=True)
```

```python
X_one_encoded
```

```python
X_one_encoded = X_one_encoded.astype(int)
```

```python
X_one_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()

column = ['model', 'transmission', 'fuelType']

XLabel = X
 
for i in column:
    XLabel[i] = encode.fit_transform(XLabel[i])
```

```python
XLabel
```

```python
XLabel['model'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

numberical_col=['year' ,'mileage', 'tax', 'mpg', 'engineSize']

scaler = StandardScaler()
X_one_encoded[numberical_col] = scaler.fit_transform(X_one_encoded[numberical_col])
```

```python
X_one_encoded
```

```python
XLabel.columns
```

```python
XLabel_col = ['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']
XLabel[XLabel_col]= scaler.fit_transform(XLabel[XLabel_col])
```

```python
XLabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encoded, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1))/(n - p - 1)
adjusted_r2
```

```python
X_train, X_test, y_train, y_test = train_test_split(XLabel, y, test_size=0.33, random_state=42)
```

```python
model2 = LinearRegression()
model2.fit(X_train, y_train)
```

```python
y_pred2 = model2.predict(X_test)
```

```python
y_pred2
```
